## **Addressing the missing values in the dataset.**


We will replace the null values in the 'rating,' 'cast,' 'country,' and 'director' columns with the string 'N/A.' This approach helps to standardize and signify the absence of specific information in these columns. The use of 'N/A' serves as a placeholder for missing values and allows for more consistent data representation during subsequent analysis.

```python
data.fillna({'rating': 'N/A', 'cast' : 'N/A', 'country': 'N/A', 'director' : 'N/A'}, inplace = True)
```

```python
data.isna().sum()
```

Now, we will handle the missing values in the 'date_added' column. Various imputation methods can be employed to address these null values. Three potential approaches are outlined:

**Manual Replacement:** Substitute missing values with actual data acquired through a Google search.

**Row Deletion:** Remove rows containing missing dates. However, this is applicable only if the number of missing values is minimal compared to the dataset size.

**Mode Imputation:** Replace missing dates with the most frequently occurring date in the 'date_added' column.

For our further analysis, we will proceed with the third option, using mode imputation to fill in the missing values in the 'date_added' column.

```python
data[data['date_added'].isnull()].head()
```

```python
frequent_occuring_date = data['date_added'].max()
frequent_occuring_date
```

```python
data.fillna({'date_added': frequent_occuring_date}, inplace = True)
```

```python
data[data['duration'].isnull()]
```


Additional data cleaning is required as there are three null values in the 'duration' column. We have two options for addressing this issue:

* Replace null values with the mean duration.

However, upon further observation, it is noted that actual duration data is erroneously placed in the 'rating' column. To rectify this, we will shift this data forward into the 'duration' column, resolving the misplaced values.

```python
duration_missing_value_index = data[data['duration'].isnull()].index
duration_missing_value_index
```

```python
data.loc[duration_missing_value_index,  'duration'] = data['rating']

```

# **Outlier check**

Outliers in a dataset are data points that differ significantly from other observations. They are often important insights signifying key events.

**Methods**: There are 1000s of outlier detection methods. The ones I use can be broken into 4 categories: Statistical, Clustering, Time Series, and Machine Learning.

**Statistical:** My favorite is the Boxplot / IQR method. A box plot is a graphical display that can help in identifying outliers. Data points that fall outside of the whiskers (typically 1.5 times the interquartile range from the box) are potential outliers. Extreme outliers are 3X IQR.

**Proximity Based:** These methods use a distance calculation. K-nearest neighbors (KNN) is commonly used to flag data points as outliers based on how isolated they are from their neighbors.

**Time Series:** Time series adds an extra layer of complexity due to trend and seasonal effects. STL (Seasonal Trend Loess) Decomposition can be used to remove these effects first. Then statistical techniques like IQR or student's t-test can be used to locate outliers.

**Machine Learning:** Specialized ML algorithms like Isolation Forest can be used to detect outliers. Isolation Forest is an ensemble method that isolates anomalies instead of profiling normal data points.

We will employ statistical methods to detect outliers in the data.

* **To identify outliers in the 'duration' column.**

```python
# Calculate Q1, Q3, and IQR for 'duration'
Q1_duration = data['duration1'].quantile(0.25)
Q3_duration = data['duration1'].quantile(0.75)
IQR_duration = Q3_duration - Q1_duration

# Define lower and upper thresholds for outliers
lower_threshold_duration = Q1_duration - 1.5 * IQR_duration
upper_threshold_duration = Q3_duration + 1.5 * IQR_duration

# Identify potential outliers in 'duration'
outliers_duration = data[(data['duration1'] < lower_threshold_duration) | (data['duration1'] > upper_threshold_duration)]

# Print or further analyze the potential outliers
outliers_duration.head()
```

As observed, there are two values in the 'duration' column that qualify as outliers.

```python
# Creating a boxplot for 'duration' to visualize the outliers.
plt.figure(figsize=(8, 4))
sns.boxplot(x='duration1', data=data, palette = 'Greens', flierprops=dict(markerfacecolor='r', marker='D'))
plt.title('Boxplot of Duration')
plt.xlabel('Duration (minutes)')
plt.show()
```

As visible in the boxplot, there are two values identified as outliers in the 'duration' column.
